```python
#print colors in set1 but not in set2
set1=set(input("enter the colors: ").split(","))
set2=set(input("enter colors: ").split(","))
print("the set 1 is:",set1)
print("the set 2 is:",set2)
diff=set1-set2
print("the colors in set1 but not in set2 is: ",diff)
```

```python
#union of colors from 2 set
set1=set(input("enter the colors: ").split(","))
set2=set(input("enter colors: ").split(","))
print("the set 1 is:",set1)
print("the set 2 is:",set2)
union=set1.union(set2)
print("the colors in set1 and set2 combined is: ",union)
```
