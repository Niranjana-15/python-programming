```python
#get the third last elements of a tuple
tuples=(1,2,3,4,5,6)
print("3rd element from the first is",tuples[2],"and 3rd element from the last:",tuples[-3])
```

```python
#create a tuple of numbers and find it's average
number=input("enter the numbers:")
tuples=tuple(map(int,number.split(",")))
print("Tuple is:",tuples)
print("average is:",sum(tuples)/len(tuples))
```
