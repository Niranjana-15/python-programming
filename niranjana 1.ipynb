```python
#program that accepts an integer and computes the value of n+nn+nnn
n=input("Enter the value of n:")
nn=n+n
nnn=n+n+n
sum=int(n)+int(nn)+int(nnn)
print(n,"+",nn,"+",nnn,"=",sum)
```
