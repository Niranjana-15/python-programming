```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)


# Getting input from the user
l1 = int(input("Enter length of Rectangle 1: "))
b1 = int(input("Enter breadth of Rectangle 1: "))

l2 = int(input("Enter length of Rectangle 2: "))
b2 = int(input("Enter breadth of Rectangle 2: "))

# Creating objects
r1 = Rectangle(l1, b1)
r2 = Rectangle(l2, b2)

print("\nRectangle 1")
print("Area:", r1.area())
print("Perimeter:", r1.perimeter())

print("\nRectangle 2")
print("Area:", r2.area())
print("Perimeter:", r2.perimeter())

# Comparing areas
if r1.area() > r2.area():
    print("Rectangle 1 has greater area")
elif r1.area() < r2.area():
    print("Rectangle 2 has greater area")
else:
    print("Both rectangles have equal area")
```

```python
#Bank Account Class:Deposit,Withdraw and Display Account Details.
class Bank:
    def __init__(self,name,account_no,type,balance):
        self.name=name
        self.account_no=account_no
        self.type=type
        self.balance=balance
    def deposit(self):
        amount = int(input("Deposit amount: "))
        print("Initial balance:",self.balance)
        self.balance+=amount
        print(amount,"is deposited")
        print("Current balance:",self.balance)
    def withdraw(self):
        amount=int(input("Withdraw amount: "))
        print("Initial balance:",self.balance)
        if amount<=self.balance:
            self.balance-=amount
            print(amount,"is withdrawn")
            print("Current balance:", self.balance)
        else:
            print("You don't have enough balance to withdraw 4!")
            print("Current balance:", self.balance)
    def account_info(self):
        print("Account holder Name:",self.name)
        print("Account Number:",self.account_no)
        print("Account Type:",self.type)
        print("Account Balance:",self.balance)

name=input("Enter name: ")
account_no=input("Enter account number: ")
type=input("Enter account type: ")
balance=int(input("Enter initial balance: "))
holder=Bank(name, account_no, type, balance)

while True:
    print("\n1. Deposit 2. Withdraw 3. Account information 4. Exit")
    choice=int(input("Choose your option: "))
    if choice==1:
        holder.deposit()
    elif choice==2:
        holder.withdraw()
    elif choice==3:
        holder.account_info()
    elif choice==4:
        print("Thank you!")
        break
    else:
        print("Invalid option")
```

```python
#Multilevel Inheritance:-Python Book Details.
class Publisher:
    def __init__(self,pname):
        self.pubname=pname
    def display1(self):
        print("\nBook details")
        print("Publisher name=",self.pubname)

class Book(Publisher):
    def __init__(self,pname,bname,author):
        self.title=bname
        self.auth=author
        Publisher.__init__(self,pname)
    def display2(self):
        print("Title=",self.title)
        print("Author=",self.auth)

class Python(Book):
    def __init__(self,pname,title,au,price,np):
        self.rate=price
        self.nop=np
        Book.__init__(self,pname,title,au)
    def display3(self):
        print("Rate=",self.rate)
        print("No.of pages=",self.nop)

p=input("Enter publisher name:")
b=input("Enter book name:")
a=input("Enter author name:")
r=int(input("Enter rate of book:"))
n=int(input("Enter no.of pages:"))
obj=Python(p,b,a,r,n)
obj.display1()
obj.display2()
obj.display3()
```

```python
#Operator Overloading:-Adiition of Two Time Object.
class Time:
    def __init__(self,hr,mint,sec):
        self.__h=hr
        self.__m=mint
        self.__s=sec
    def __add__(self,other):
        hh=self.__h+other.__h
        mm=self.__m+other.__m
        ss=self.__s+other.__s
        if ss>=60:
            m=int(ss/60)
            ss=ss%60
            mm+=m
        if mm>=60:
            h=int(mm/60)
            mm=mm%60
            hh+=h
        print("Added time..")
        print("Hour:",hh,"Minute:",mm,"Second:",ss)

h1=int(input("Enter hour1:"))
m1=int(input("Enter minute1:"))
s1=int(input("Enter second1:"))

h2=int(input("Enter hour2:"))
m2=int(input("Enter minute2:"))
s2=int(input("Enter second2:"))

t1=Time(h1,m1,s1)
t2=Time(h2,m2,s2)
t1+t2
```
