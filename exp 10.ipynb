```python
#Finding the number of times the word repeated
text = input("Enter a line of sentence:")
print("The sentence is:",text)
words = text.split()
for word in words:
    print(word, ":", words.count(word))
```

```python
#store the number greater than 100 as 100 and other as a list
numbers=input("Enter integers separated by space: ").split()
print("the numbers are:",numbers)
for i in range(len(numbers)):
    numbers[i]=int(numbers[i])
for i in range(len(numbers)):
    if numbers[i]>100:
        numbers[i]="over"
print("The list is:",numbers)
```

```python
#finding the no. of times the lettera repeated in names
names= input("Enter first name separated by space: ").split()
print("The names are:",names)
count=0
for name in names:
    count=count+name.lower().count("a")
print("The number of letters a in the names is:",count)
```

```python
#finding the gcd of a 2 number
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
while b!=0:
    a,b=b,a%b
print("GCD is:", a)
```

```python
#List of numbers after removing even numbers
numbers=input("Enter the values separated by commas:").split(",")
print("The number are:",numbers)
new=[]
for i in numbers:
    i=int(i)
    if i % 2 != 0:
        new.append(i)
print("List after removing even numbers:",new)
```

```python
#factorial of a number
n=int(input("Enter a number: "))
fact=1
for i in range(1,n+1):
    fact=fact*i
print("Factorial is:",fact)
```

```python
#Fibonacci of n numbers
n=int(input("Enter the number of terms: "))
a=0
b=1
for i in range(n):
    print(a,end=",")
    c=a+b
    a=b
    b=c
```

```python
#4 digit even number and also perfect square
start=int(input("Enter the starting number: "))
end=int(input("Enter the ending number: "))
print("The strain number is",start,"and the ending number is:",end)
result=[]
for num in range(start,end+1):
    if 1000<=num<= 9999:
        if int(num ** 0.5) ** 2==num:
            a=num // 1000
            b=(num // 100)%10
            c=(num // 10)%10
            d=num%10
            if a%2==0 and b%2==0 and c%2==0 and d%2==0:
                result.append(num)
print("The numbers are:",result)
```

```python
#count the number of characters
strings=input("Enter the string:")
print("The string is:",strings)
frequency={}
for char in strings:
    frequency[char] = frequency.get(char, 0) + 1
print("Character frequency:")
for char, count in frequency.items():
    print(char, ":", count)
```

```python
#check if the list length and sum are equal
list1 = list(map(int, input("Enter first list of integers: ").split(",")))
print("The list 1 is:",list1)
list2 = list(map(int, input("Enter second list of integers: ").split(",")))
print("The list 2 is:",list2)
if len(list1)==len(list2):
    print("(a) Both lists have the same length.")
else:
    print("(a) Lists have different lengths.")
if sum(list1)==sum(list2):
    print("(b) Both lists have the same sum.")
else:
    print("(b) Lists have different sums.")
if set(list1).intersection(set(list2)):
    print("(c) Common value(s):",set(list1).intersection(set(list2)))
else:
    print("(c) No common values.")
    
```

```python
#multiples of the row number in pyramid
n=int(input("Enter the step number: "))
print("The step numbers are:",n)
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i * j,end=" ")
    print()
```
