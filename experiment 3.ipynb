```python
#write a python program to change the character first and last
s=input("enter a string:")
new=s[-1]+s[1:-1]+s[0]
print("new string:",new)
```

```python
#get a string made of the first 2 and the last 2 chars 
s=input("enter a string:")
new=s[:2]+s[-2:]
print("new string:",new)
```

```python
#create a single string seperated with space from 2 strings by swapping chars
s1=input("enter first string:")
s2=input("enter second string:")
new_s1=s2[0]+s1[1:]
new_s2=s1[0]+s2[1:]
print(new_s1+""+new_s2)
```

```python
#string from a given string where all occurences of its first char changed to $
s=input("enter a string:")
new=s[0]+s[1:].replace(s[0],'$')
print("new string:",new)
```
