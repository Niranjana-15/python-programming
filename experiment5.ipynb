```python
#number seperated with commas stored in tuple
number=input("enter the numbers seperated with commas:")
print("the numbers are:",number)
num_t=tuple(int(x) for x in number.split(","))
average=sum(num_t)/len(num_t)
print("the average is:",average)
```
