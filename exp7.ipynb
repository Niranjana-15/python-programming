```python
#print colors in set1 but not in set2
a=set(input("enter the colors: ").split(","))
b=set(input("enter colors: ").split(","))
colors=a-b
print("the colors are: ",colors)
```

```python
#union of colors from 2 set
a=set(input("enter the colors: ").split(","))
b=set(input("enter colors: ").split(","))
colors=a|b #represet union
print("the colors are: ",colors)
```
